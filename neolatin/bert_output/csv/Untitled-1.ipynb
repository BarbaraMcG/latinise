```python
import pandas as pd

# --- Step 1: Load the CSVs ---

t0 = pd.read_csv("neighbors_t0_finetuned_no-lem.csv")

t1 = pd.read_csv("neighbors_t1_finetuned_no-lem.csv")

t2 = pd.read_csv("neighbors_t2_finetuned_no-lem.csv")

t3 = pd.read_csv("neighbors_t3_finetuned_no-lem.csv")
```

```python
# --- Step 2: Clean and keep only lemma + neighbour ---

# Assumes the CSVs have columns like: lemma, neighbour, cosine

# If column names differ, adjust below

t0 = t0[['Target Word', 'Neighbor', 'Cosine Similarity']]

t1 = t1[['Target Word', 'Neighbor', 'Cosine Similarity']]

t2 = t2[['Target Word', 'Neighbor', 'Cosine Similarity']]

t3 = t3[['Target Word', 'Neighbor', 'Cosine Similarity']]
```

```python
# Combine all DataFrames into one
all_neighbors = pd.concat([t0, t1, t2, t3], ignore_index=True)

# Choose the lemma you want to search for, e.g., "torcular"
lemma = "torcular"

# Filter and print all rows with that lemma
print(all_neighbors[all_neighbors['Target Word'] == lemma])
```

```python
# --- Step 2: Clean and keep only lemma + neighbour + cosine similarity ---

# Select relevant columns and add provenance label for each time slice
t0 = t0[['Target Word', 'Neighbor']].copy()
t0['Time Slice'] = 't0'

t1 = t1[['Target Word', 'Neighbor']].copy()
t1['Time Slice'] = 't1'

t2 = t2[['Target Word', 'Neighbor']].copy()
t2['Time Slice'] = 't2'

t3 = t3[['Target Word', 'Neighbor']].copy()
t3['Time Slice'] = 't3'

# Combine all DataFrames into one
all_neighbors = pd.concat([t0, t1, t2, t3], ignore_index=True)

# Choose the lemma you want to search for, e.g., "torcular"
lemma = "sacerdos"

# Filter and print all rows with that lemma, including provenance
print(all_neighbors[all_neighbors['Target Word'] == lemma])
```
